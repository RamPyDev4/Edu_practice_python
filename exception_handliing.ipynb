```python

def function_c():
    print("C: Entered function C")

    print("C: About to raise exception")
    raise ValueError("Error from C")

    print("C: After exception")  # Will this execute?


def function_b():
    print("B: Entered function B")

    print("B: Calling function C")
    function_c()

    print("B: Returned from function C")  # Will this execute?


def function_a():
    print("A: Entered function A")

    try:
        print("A: Calling function B")
        function_b()

        print("A: Returned from function B")

    except ValueError as exc:
        print("A: Exception handled:", exc)

    print("A: Function A completed")


print("Program started")
function_a()
print("Program completed")
```

```python
def process_number(num):
    print("Processing:", num)

    if num == 3:
        raise ValueError("Three is not allowed")

    print("Completed:", num)


numbers = [1, 2, 3, 4, 5]

for num in numbers:
    try:
        process_number(num)

    except ValueError as exc:
        print("Error:", exc)

print("All processing finished")
```

```python
print("1. Program started")


class Employee:

    print("2. Employee class is being defined")

    def __init__(self, name, age):

        print("\n3. __init__() started")

        print("4. Value received in name:", name)
        print("5. Value received in age:", age)

        print("6. Current object (self):", self)

        self.name = name
        self.age = age

        print("7. Stored name:", self.name)
        print("8. Stored age:", self.age)

        print("9. __init__() completed")


print("\n10. Creating first employee")

emp1 = Employee("Mantesh", 30)

print("\n11. First object created:", emp1)

print("\n12. Creating second employee")

emp2 = Employee("Ravi", 28)

print("\n13. Second object created:", emp2)

print("\n14. Program completed")
```
