```python
n = 3
while n > 0:
    print(n)
    n -= 1
```

```python
nums = [2, 5, 1]
i = 0
total = 0
while i < len(nums):
    total += nums[i]
    i += 1
print(total)
```

```python
n = 4

for i in range(1, n + 1):
    print("*" * i)
```

```python
n = 5

for i in range(n, 0, -2):
    print("*" * i)
```

```python
n = 3

for _ in range(n):
    print("*" * n)
```

```python
n = 5

for i in range(n):
    if i == 0 or i == n - 1:
        print("*" * n)
    else:
        print("*" + " " * (n - 2) + "*")
```

```python
n = 4

for i in range(1, n + 1):   
    for j in range(1, i + 1): 
        print(j, end="")
    print()
```
