```python
### 6. Floyd’s triangle (continuous numbers)
n = 4
num = 1
for i in range(1, n + 1):
    for _ in range(i):
        print(num, end=" ")
        num += 1
    print()
```

```python
### 7. Centered star pyramid
n = 4
for i in range(1, n + 1):
    spaces = " " * (n - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
```

```python
# 8. Inverted centered star pyramid
n = 4
for i in range(n, 0, -1):
    spaces = " " * (n - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
```

```python
# 9. Diamond (centered, stars)
n = 4
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
# 10. Same digit per row
n = 4
for i in range(1, n + 1):
    print((str(i) + " ") * i)
```

```python
# 11. Binary triangle (0 and 1)
n = 5
val = 1
for i in range(1, n + 1):
    for _ in range(i):
        print(val, end="")
        val = 1 - val
    print()
```

```python
# 12. Checkerboard
size = 4
for r in range(size):
    for c in range(size):
        cell = "#" if (r + c) % 2 == 0 else "."
        print(cell, end=" ")
    print()
```

```python
# 13. Hollow diamond
n = 4
for i in range(n):
    left = n - i - 1
    mid = 2 * i - 1
    if i == 0:
        print(" " * left + "*")
    else:
        print(" " * left + "*" + " " * mid + "*")
for i in range(n - 2, -1, -1):
    left = n - i - 1
    mid = 2 * i - 1
    if i == 0:
        print(" " * left + "*")
    else:
        print(" " * left + "*" + " " * mid + "*")        
```

```python
# 14. Alphabet triangle
n = 4
for i in range(1, n + 1):
    for j in range(i):
        print(chr(ord("A") + j), end="")
    print()

```

```python
# 15. Centered number pyramid
n = 4
for i in range(1, n + 1):
    nums = "".join(str(x) for x in range(1, i + 1))
    line = nums + nums[-2::-1] if i > 1 else nums
    pad = " " * (n - i)
    print(pad + line)
```
