```python
### 6. Floyd’s triangle (continuous numbers)
n = 4
num = 1
for i in range(1, n + 1):
    for _ in range(i):
        print(num, end=" ")
        num += 1
    print()
```

```python
### 7. Centered star pyramid
n = 4
for i in range(1, n + 1):
    spaces = " " * (n - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
```

```python
# 8. Inverted centered star pyramid
n = 4
for i in range(n, 0, -1):
    spaces = " " * (n - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
```

```python
# 9. Diamond (centered, stars)
n = 4
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
# 10. Same digit per row
n = 4
for i in range(1, n + 1):
    print((str(i) + " ") * i)
```

```python
# 11. Binary triangle (0 and 1)
n = 5
val = 1
for i in range(1, n + 1):
    for _ in range(i):
        print(val, end="")
        val = 1 - val
    print()
```

```python
# 12. Checkerboard
size = 4
for r in range(size):
    for c in range(size):
        cell = "#" if (r + c) % 2 == 0 else "."
        print(cell, end=" ")
    print()
```

```python
# 13. Hollow diamond
n = 4
for i in range(n):
    left = n - i - 1
    mid = 2 * i - 1
    if i == 0:
        print(" " * left + "*")
    else:
        print(" " * left + "*" + " " * mid + "*")
for i in range(n - 2, -1, -1):
    left = n - i - 1
    mid = 2 * i - 1
    if i == 0:
        print(" " * left + "*")
    else:
        print(" " * left + "*" + " " * mid + "*")        
```

```python
# 14. Alphabet triangle
n = 4
for i in range(1, n + 1):
    for j in range(i):
        print(chr(ord("A") + j), end="")
    print()

```

```python
# 15. Centered number pyramid
n = 4
for i in range(1, n + 1):
    nums = "".join(str(x) for x in range(1, i + 1))
    line = nums + nums[-2::-1] if i > 1 else nums
    pad = " " * (n - i)
    print(pad + line)
```

```python
##16. Right-aligned number triangle
n = 4
for i in range(1, n + 1):
    nums = "".join(str(x) for x in range(1, i + 1))
    print(nums.rjust(n + i - 1))
```

```python
# 17. Butterfly (stars, two wings)
n = 3
for i in range(1, n + 1):
    print("*" * i + " " * (2 * (n - i)) + "*" * i)
for i in range(n, 0, -1):
    print("*" * i + " " * (2 * (n - i)) + "*" * i)
```

```python
# 18. Hourglass (stars)
n = 5
for i in range(n, 0, -2):
    print(" " * ((n - i) // 2) + "*" * i)
for i in range(3, n + 1, 2):
    print(" " * ((n - i) // 2) + "*" * i)
```

```python
# 19. Pascal’s triangle (coefficients)
rows = 5
prev = []
for r in range(rows):
    cur = [1]
    for j in range(1, r):
        cur.append(prev[j - 1] + prev[j])
    if r > 0:
        cur.append(1)
    print(" ".join(str(x) for x in cur).center(rows * 3))
    prev = cur
```

```python
# 20. X pattern (diagonals)
n = 5
for r in range(n):
    for c in range(n):
        if r == c or r + c == n - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# 21. Plus pattern (cross)
n = 5
mid = n // 2
for r in range(n):
    for c in range(n):
        if r == mid or c == mid:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# 22. Number grid (row × column values)
n = 4
for r in range(1, n + 1):
    for c in range(1, n + 1):
        print(r * c, end="\\t")
    print()
```

```python
# 23. Hollow rectangle
rows, cols = 3, 6
for r in range(rows):
    if r == 0 or r == rows - 1:
        print("*" * cols)
    else:
        print("*" + " " * (cols - 2) + "*")
```

```python
# 24. Inverted number triangle (all digits to n)
n = 4
for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

```python
# 25. Zigzag (stars on a shifting diagonal)
cols = 7
for r in range(3):
    line = [" "] * cols
    for c in range(cols):
        if (c + r) % 3 == 0:
            line[c] = "*"
    print("".join(line))
```

```python
# 26. GCD stripes (decorative bands)
h, w = 4, 12
for r in range(h):
    for c in range(w):
        print("=" if (r + c) % 3 == 0 else "-", end="")
    print()
```

```python
# 27. Staircase blocks (#)
n = 4
for i in range(1, n + 1):
    print(("#" * i).ljust(n) + " " + ("#" * i))
```

```python
# 28. Descending triangle of consecutive numbers
n = 4
start = n * (n + 1) // 2
for i in range(n, 0, -1):
    row = []
    for _ in range(i):
        row.append(str(start))
        start -= 1
    print(" ".join(row))
```

```python
# 29. Square border with digits (interior empty)
n = 5
for r in range(n):
    for c in range(n):
        if r == 0 or r == n - 1 or c == 0 or c == n - 1:
            print((r + c) % 10, end="")
        else:
            print(" ", end="")
    print()
```

```python
# 30. Arrow head (right-pointing)
n = 5
for i in range(1, n + 1, 2):
    print("*" * i)
for i in range(n - 2, 0, -2):
    print("*" * i)
```
